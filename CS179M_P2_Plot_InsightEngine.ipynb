```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
filename = "pecan1212.txt"
k = 3
data = np.loadtxt(filename)
def distance(p1, p2):
    return np.sqrt((p1[0]-p2[0])**2 + (p1[1]-p2[1])**2)
def nearest_neighbor_order(points):
    n = len(points)
    if n == 0:
        return []
    unvisited = list(range(1, n))
    route = [0]
    while unvisited:
        last = route[-1]
        next_city = min(unvisited, key=lambda j: distance(points[last], points[j]))
        route.append(next_city)
        unvisited.remove(next_city)
    return route
kmeans = KMeans(n_clusters=k, n_init=10, random_state=42).fit(data)
labels = kmeans.labels_
plt.figure(figsize=(7, 7))
plt.scatter(data[:, 0], data[:, 1], c="lightgray", s=15, label="Locations")
colors = ["tab:orange", "tab:blue", "tab:green", "tab:red"]
route_distances = []
for i in range(k):
cluster_points = data[labels == i]
pad = np.mean(cluster_points, axis=0)
order = nearest_neighbor_order(cluster_points)
ordered_points = cluster_points[order]
route_points = np.vstack([pad, ordered_points, pad])
dist = 0.0
for j in range(len(route_points) - 1):
        dist += distance(route_points[j], route_points[j+1])
route_distances.append(dist)
plt.plot(
route_points[:, 0],
route_points[:, 1],
linestyle="-",
marker="o",
markersize=4,
color=colors[i],
        alpha=0.9,
        label=f"Drone {i+1} (min {dist/100:.1f})"
    )
max_route = max(route_distances)
flight_time = max_route / 100.0
setup_time = 2 * k
time_in_field = flight_time + setup_time
plt.title(f"Overall Solution for K={k} | Time in Field: {time_in_field:.1f} min")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True, linestyle="--", alpha=0.4)
plt.axis("equal")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
filename = "pecan1212.txt"
k = 3                        # number of drones (1–4)
data = np.loadtxt(filename)
def distance(p1, p2):
    return np.sqrt((p1[0]-p2[0])**2 + (p1[1]-p2[1])**2)
def nearest_neighbor_order(points):
    """Return NN visiting order of indices for 'points' (0-based)."""
    n = len(points)
    if n == 0:
        return []

    unvisited = list(range(1, n))
    route = [0]

    while unvisited:
        last = route[-1]
        next_city = min(unvisited, key=lambda j: distance(points[last], points[j]))
        route.append(next_city)
        unvisited.remove(next_city)

    return route
kmeans = KMeans(n_clusters=k, n_init=10, random_state=42).fit(data)
labels = kmeans.labels_
plt.figure(figsize=(7, 7))
plt.scatter(data[:, 0], data[:, 1], c="lightgray", s=15, label="Locations")
colors = ["tab:orange", "tab:blue", "tab:green", "tab:red"]
route_distances = []
for i in range(k):
    cluster_points = data[labels == i]
    pad = np.mean(cluster_points, axis=0)
    order = nearest_neighbor_order(cluster_points)
    ordered_points = cluster_points[order]
    route_points = np.vstack([pad, ordered_points, pad])
    dist = 0.0
    for j in range(len(route_points) - 1):
        dist += distance(route_points[j], route_points[j+1])
    route_distances.append(dist)
    plt.plot(
        route_points[:, 0],
        route_points[:, 1],
        linestyle="-",
        marker="o",
        markersize=4,
        color=colors[i],
        alpha=0.9,
        label=f"Drone {i+1} (min {dist/100:.1f})"
    )
max_route = max(route_distances)
flight_time = max_route / 100.0
setup_time = 2 * k
time_in_field = flight_time + setup_time

plt.title(f"Overall Solution for K={k} | Time in Field: {time_in_field:.1f} min")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True, linestyle="--", alpha=0.4)
plt.axis("equal")
plt.legend()
plt.tight_layout()
plt.show()
```
